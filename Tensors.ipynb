Time Measurement between CPU and GPU

```python
import time
```

```python
time.time()
```

```python
size=1000

# torch.manual_seed(10)  ##// here 10 is seed value which define starting point, it could be any value
mat_1=torch.rand(size,size)
mat_2=torch.rand(size,size)

print(mat_1)
print(mat_2)

## Measure time on CPU
start_time=time.time()

mat_mul=torch.matmul(mat_1, mat_2)

cpu_time=time.time() - start_time

print(f"Matrix Multiplication time on CPU {cpu_time: .4f} seconds")

## Moving to GPU

mat_1=mat_1.to(device)
mat_2=mat_2.to(device)

## Measure time on GPU

## Measure time on CPU
start_time=time.time()

mat_mul=torch.matmul(mat_1, mat_2)
torch.cuda.synchronize() # ensure all GPU operations are completed

GPU_time= time.time() - start_time

print(f"Matrix Multiplication time on GPU {GPU_time: .4f} seconds")
```
